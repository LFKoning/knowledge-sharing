```python
import numpy as np
import pandas as pd

import helpers as h

%load_ext autoreload
%autoreload 2
```

```python
# Default image properties
properties = {
    "width": 1000,
    "height": 250,
}
```

```python
# Generate a time series with a linear trend and seasonality.
np.random.seed(1)

n = 75
intercept = 10
slope = 2
seasonality = [29, 30, 25, 27, 31, 43, 37]

y = h.generate_linear(n, slope=slope, intercept=intercept)
y = h.add_seasonality(y, seasonality, 2)
y = h.add_noise(y, sd=3)

h.plot_timeseries(y, properties=properties).configure(padding=30)
```

```python
X = pd.DataFrame({
    "time": y.index,
    "day_1": (y.index % 7 == 1).astype(int),
    "day_2": (y.index % 7 == 2).astype(int),
    "day_3": (y.index % 7 == 3).astype(int),
    "day_4": (y.index % 7 == 4).astype(int),
    "day_5": (y.index % 7 == 5).astype(int),
    "day_6": (y.index % 7 == 6).astype(int),
})

X.head(3)
```

```python
from sklearn.linear_model import LinearRegression

reg = LinearRegression()
reg.fit(X, y)
```

```python
pd.DataFrame(zip(X.columns, reg.coef_), columns=["Variable", "Coefficient"])
```

```python
ahead = pd.RangeIndex(start=n+1, stop=n+20)
X_ahead = pd.DataFrame({
    "time": ahead,
    "day_1": (ahead % 7 == 1).astype(int),
    "day_2": (ahead % 7 == 2).astype(int),
    "day_3": (ahead % 7 == 3).astype(int),
    "day_4": (ahead % 7 == 4).astype(int),
    "day_5": (ahead % 7 == 5).astype(int),
    "day_6": (ahead % 7 == 6).astype(int),
}, index=ahead)
```

```python
reg.predict(X_ahead)
```

```python
h.plot_timeseries(y, colors="blue") + h.plot_timeseries(pd.Series(reg.predict(X_ahead), index=ahead), colors="orange")
```
